# Order Analysis
This notebook loads order data and produces a monthly summary.

```python
import pandas as pd
import json
from snowflake.snowpark.context import get_active_session
import os

session = get_active_session()
print('Session created')
```

```python
df = session.table('AGG_ORDERS').to_pandas()
print(f'loaded {len(df)} rows')
```

```python
summary = df.groupby(['CUSTOMER_ID','order_year']).agg(total_amount=('total_amount','sum'),order_count=('order_count','sum')).reset_index()
summary.head()
```
